```python
# Function to print a full name multiple times
def name(firstname, lastname):
    # Initialize a counter variable 'i' to 0
    i = 0
    # Use a while loop to repeat the printing of the full name
    while i < 3:  # Loop will run 3 times
        # Print the first name and last name
        print(firstname, lastname)
        # Increment the counter by 1 after each iteration
        i += 1

# Call the function with the first name "nihal" and last name "mishra"
name("nihal", "mishra")
```

```python
#function to calculate factorial
def factorial(n):#function definition
    if n==0:#if n is 0
        return 1#return 1
    else:#if n is not 0
        return n*factorial(n-1)#return n*factorial(n-1)
factorial(5)#calling the function to calculate factorial of 5
```

```python
def calc_fac(n):#function to calculate factorial
    fact=1#initializing fact to 1
    for i in range(1,n+1):#iterating through the range
        fact=fact*i#calculating factorial    
    return fact#returning the factorial
calc_fac(6)#calling the function to calculate factorial of 6
```

```python
# Function to divide two numbers and print the result
def divide(a, b):
    # Perform the division of 'a' by 'b'
    division = a / b
    # Print the result in a formatted manner
    print(a, "/", b, "=", division)

# Call the function with two numbers: 10 divided by 5
divide(10, 5)

# Call the function with two numbers: 100 divided by 5
divide(100, 5)


```

```python
# Function to generate and print the multiplication table for a given number
def table(a):
    # Loop through numbers from 1 to 10 to create the multiplication table
    for i in range(1, 11):
        # Print the current multiplication result
        print(a, "*", i, "=", a * i)

# Call the function to execute the multiplication table logic
table(5)



```

```python
def calculator():
    # Function to calculate the sum of two numbers
    def sum(a, b):
        print("Sum:", a + b)
    
    # Function to calculate the subtraction of two numbers
    def sub(a, b):
        print("Subtraction:", a - b)
    
    # Function to calculate the multiplication of two numbers
    def multiply(a, b):
        print("Multiplication:", a * b)
    
    # Function to calculate the division of two numbers
    def divide(a, b):
        # Check if the denominator is not zero to avoid division errors
        if b != 0:
            print("Division:", a / b)
        else:
            print("Cannot divide by zero")
    
    # Return a dictionary of operations to allow flexible usage
    return {"sum": sum, "sub": sub, "multiply": multiply, "divide": divide}

# Usage of the calculator functions
calc = calculator()                 # Initialize the calculator
calc["sum"](100, 200)               # Call the sum function with 100 and 200
calc["sub"](200, 50)                # Call the subtraction function with 200 and 50
calc["multiply"](10, 20)            # Call the multiplication function with 10 and 20
calc["divide"](100, 2)              # Call the division function with 100 and 2
```
