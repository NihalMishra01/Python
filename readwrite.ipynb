<h1 align="center" style="color: black; font-family: 'black'; text-decoration: underline; font-weight: bold; background-color: white;">READING FILE</h1>


```python
f=open( "nihal.txt","r")#opening file

data=f.read()#reading

print(data)#printing data
print(type(data))#printing type

f.close()#closing file
```

```python
f=open( "nihal.txt","r")#Opening file 
data=f.readline(20)#reading line
print(data)#printing first 5 characters from the line
f.close()#closing file
```

```python
f=open("nihal.txt","r")#opening a file

line1=f.readline()#reading line1
print(line1)#printing line

line2=f.readline()#reading line2
print(line2)#printing line

f.close()
```

```python
f = open("nihal.txt", "r")  # Opening file

lines = f.readlines()  # Reading all lines into a list

print(lines)  # Printing list of lines
print(type(lines))  # Printing type (list)

f.close()  # Closing file
```

```python
with open("nihal.txt", "r") as f:  # Opening file using with
    data = f.read()  # Reading file content
    
print(data)  # Printing data
```

<h1 align="center" style="color: black; font-family: 'black'; text-decoration: underline; font-weight: bold; background-color: white;">WRITING IN FILE</h1>

```python
f1=open("nihal.txt","w")#opening file in write mode
f1.write("hi ths is a new line ")#writing data  or removing existing data 
f1.close()#closing the file

f2=open("nihal.txt","a")#opening file in append mode

f2.write("\nthis is another line")#Appending data

f2.close()#closing the file

f3=open("nihal.txt","r")
data=f3.read()
print(data)


```
