DICTIONARY & SET IN PYTHON

```python
information = {
    'name': 'NIHAl MISHRA',
    'age': 19,
    'course': 'BTECH CSE WITH SPECIALIZATION IN ARTIFICIAL INTELLIGENCE AND DATA SCIENCE',
    'Semester': 4,
    '3rd semester SGPA': 8.86,
    'Learning skills': ('PYTHON', 'DATA SCIENCE', 'SQL', 'POWER BI'),
    'Aim': 'To pursue a career in Data Science and AI',
}

# Printing specific key-value pairs in a formatted way
print('Name:', information['name'],
      '\nAge:', information['age'],
      '\nCourse:', information['course'],
      '\nSemester:', information['Semester'],
      '\n3rd semester SGPA:', information['3rd semester SGPA'],
      '\nLearning skills:', information['Learning skills'],
      '\nAim:', information['Aim'],
      '\n')

# Printing all keys in the dictionary
print(information.keys())

# Printing all values in the dictionary
print(information.values())

# Printing all key-value pairs as items (tuples)
print(information.items())

# Updating the SGPA value in the dictionary
information['3rd semester SGPA'] = 8.87

# Printing the updated SGPA
print("New SGPA:", information["3rd semester SGPA"])
```

```python
# Define a dictionary with student details
student = {
    'name': 'NIHAL MISHRA',
    # Corrected: Added commas between the subjects (dictionaries need valid syntax)
    'subjects': {
        'operating system',
        'data structures',
        'python',
        'theory of computation'
    }
}
# Print the entire dictionary
print(student)
# Print all keys in the dictionary
print(student.keys())
# Print all values in the dictionary
print(student.values())
# Loop through the dictionary to print keys and values
for key, value in student.items():
    print(f"{key}: {value}")
```

```python
# Define two sets
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# Print the sets
print("Set 1:", set1)
print("Set 2:", set2)
```

```python
# Union combines elements from both sets (no duplicates)
set1 = {1, 2, 3}
set2 = {3, 4, 5}
union_set = set1.union(set2)
print("Union of Set 1 and Set 2:", union_set)
```

```python
# Intersection finds common elements between sets
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
intersection_set = set1.intersection(set2)
print("Intersection of Set 1 and Set 2:", intersection_set)
```

```python
# Difference finds elements in one set but not in the other
set1 = {1, 2, 3, 4}
set2 = {3, 4, 5, 6}
difference_set = set1.difference(set2)
print("Difference (Set 1 - Set 2):", difference_set)
```

```python
# Symmetric difference finds elements in either set, but not both
set1 = {1, 2, 3}
set2 = {3, 4, 5}
sym_diff_set = set1.symmetric_difference(set2)
print("Symmetric Difference of Set 1 and Set 2:", sym_diff_set)
```

```python
# Add a single element to the set
my_set = {1, 2, 3}
print(my_set)
my_set.add(4)
print("Set after adding 4:", my_set)
```

```python
# Remove an element from the set (raises error if element not found)
my_set = {1, 2, 3}
my_set.remove(2)
print("Set after removing 2:", my_set)
```

```python
# Discard removes an element but does NOT raise an error if not found
my_set = {1, 2, 3}
my_set.discard(4)  # 4 is not in the set, but no error
print("Set after discarding 4:", my_set)
```

```python
# Check if one set is a subset of another
set1 = {1, 2}
set2 = {1, 2, 3, 4}
print("Is Set 1 a subset of Set 2?", set1.issubset(set2))
```

```python
# Check if one set is a superset of another
set1 = {1, 2, 3, 4}
set2 = {1, 2}
print("Is Set 1 a superset of Set 2?", set1.issuperset(set2))
```

```python
# Clear removes all elements from the set
my_set = {1, 2, 3}
my_set.clear()
print("Set after clearing:", my_set)
```
