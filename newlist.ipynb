#LIST COMPREHENNSION AND LIST METHODS IN PYTHON

```python
list1=[1,2,3,4,1,5]
list2=[6,7,8,9,10]
print(list1)#print list1
print(list2)#print list2
list3=list1+list2 #concatenate list1 and list2
print(list3)#print list3
list1.append(6)#append 6 to list1
print(list1)#printing after appending 6
list1.insert(0,0)#insert 0 at the beginning of list1
print(list1)#printing after inserting 0
list1.extend([7,8,9])#extend list1 with 7,8,9
print(list1)#printing after extending
list1.remove(9)#remove 9 from list1 
print(list1)#printing after removing 9
print(list1.count(1))#count the number of 1 in list1
```

```python
list2=list1.copy()#copy list1
print(list2)
print(list1[-4])
```

```python
items=["pasta","maggie","pizza","sandwich"]
print(items)#print names
print(items[0])#print first element
items.append("dosa")#append to items
print(items)#print names after appending
items.insert(0,"burger")#insert at the beginning
print(items)#print names after inserting
items.insert(2,"idly")#insert at 2nd position
print(items[1:3])#print 2nd and 3rd element
items.extend(["fries","coke","pepsi"])#extend items
print(items)#print names after extending
```

```python
#['burger', 'pasta', 'idly', 'maggie', 'pizza', 'sandwich', 'dosa', 'fries', 'coke', 'pepsi' list items
print(len(items))#print length of items
print(items.index("burger"))#print index of burger
print(items.count("burger"))#print count of burger
items.sort()#sort items
print(items)#print sorted items
items.reverse()#reverse items
print(items)#print reversed items
"dosa" in items#check if dosa is in items
items2=items.copy()#copy items to items2
print(items2)#print items2
items3=["manchurian","noodles","momos"]#create items3
items2.extend(items3)#extend items2 with items3
print(items2)#print items2 after extending
items2.pop()#pop last element
print(items2)#print items2 after popping
items2.clear()#clear items2
print(items2)#print items2 after clearing


```

```python
if "burger" in items:#check if burger is in items
    print("yes")
else:    
    print("no")
```

```python
lst=[i*i for i in range(10)]#list comprehension
print(lst)
lst=[i for i in range(20) if i%2==0]#list comprehension
print(lst)
lst=[i for i in range(20) if i%2!=0]#list comprehension
print(lst)
```
