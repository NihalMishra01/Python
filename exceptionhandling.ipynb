# Python Exception Handling Documentation

## Description:
This script demonstrates various exception handling techniques in Python. Exception handling helps prevent a program from crashing by catching errors and allowing the program to continue execution gracefully.

## Key Concepts Covered:
1. **try-except**: Used to catch and handle specific exceptions.
2. **try-except-else**: Executes additional code if no exceptions occur.
3. **try-except-finally**: Ensures that cleanup code runs regardless of whether an exception occurs.
4. **Handling multiple exceptions**: Catching different types of errors in a single try block.
5. **Raising custom exceptions**: Creating and handling user-defined exceptions.

## Exception Types Demonstrated:
1. **ZeroDivisionError** - Occurs when dividing by zero.
2. **ValueError** - Raised when an invalid type is provided as input.
3. **IndexError** - Accessing an index out of range in a list.
4. **KeyError** - Accessing a missing key in a dictionary.
5. **FileNotFoundError** - Handling missing files.
6. **AttributeError** - Calling a method that doesn't exist.
7. **TypeError** - Using incompatible data types in operations.
8. **ModuleNotFoundError** - Importing a nonexistent module.
9. **NameError** - Using an undefined variable.
10. **RuntimeError** - Handling manually raised exceptions.

## Usage:
This documentation can be referenced when implementing exception handling in Python programs. The included code examples ensure that critical operations such as file handling, database connections, and mathematical computations are executed safely and reliably.



```python
try:
    num1 = int(input("Enter the numerator: "))
    num2 = int(input("Enter the denominator: "))
    result = num1 / num2
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except ValueError:
    print("Error: Please enter valid numbers!")
finally:
    print("Execution completed.")
```

```python
try:
    numbers = [1, 2, 3]
    index = int(input("Enter an index: "))
    print("Number at index:", numbers[index])
except IndexError:
    print("Error: Index out of range!")
except ValueError:
    print("Error: Please enter a valid number!")
```

```python
try:
    num = int(input("Enter a number: "))
    print(f"You entered: {num}")
except ValueError:
    print("Error: Invalid input! Please enter a number.")
else:
    print("No exceptions occurred!")
```

```python
try:
    with open("example.txt", "r") as file:
        content = file.read()
        print(content)
except FileNotFoundError:
    print("Error: File not found!")
```

```python
try:
    num = int(input("Enter a positive number: "))
    if num < 0:
        raise ValueError("Negative numbers are not allowed!")  # Manually raising an exception
except ValueError as ve:
    print(f"Error: {ve}")
else:
    print(f"You entered: {num}")
```

```python
try:
    student = {"name": "Nihal", "age": 19}
    print(student["grade"])  # Key doesn't exist
except KeyError:
    print("Error: The key does not exist in the dictionary!")
```

```python
try:
    num = 5 + "10"  # Adding integer and string
except TypeError:
    print("Error: Cannot add integer and string!")
```

```python
try:
    num = 10
    num.append(1)  # `int` has no `append` method
except AttributeError:
    print("Error: Object has no such attribute!")
```

```python
try:
    import nonexistentmodule  # This module does not exist
except ModuleNotFoundError:
    print("Error: Module not found!")
```

```python
try:
    print(my_variable)  # Variable not defined
except NameError:
    print("Error: Variable is not defined!")
```

```python
class CustomError(Exception):
    pass

try:
    raise CustomError("This is a custom error!")
except CustomError as e:
    print(f"Custom Exception Caught: {e}")
```

```python
try:
    file = open("nihal.txt", "r")
    content = file.read()
    print(content)
except FileNotFoundError:
    print("Error: File not found!")
finally:
    file.close()  # Ensures the file is closed even if an error occurs
    print("File closed successfully.")
```

```python
try:
    a = int(input("Enter numerator: "))
    b = int(input("Enter denominator: "))
    result = a / b
    print("Result:", result)
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except ValueError:
    print("Error: Invalid input! Enter a number.")
finally:
    print("Execution completed.")
```
