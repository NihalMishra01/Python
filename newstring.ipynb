```python
a="hey i am Nihal Mishra,Today i am learning about string methods and string slicing"
print(a)
print(a[0]) #Accessing the string using index
print(a[0:21]) #Accessing the string using slicing
print(a[4:21]) #Accessing the string using slicing
```

```python
s1="Hello"
s2="World"
print(s1+s2) #Concatenation of two strings
print(s1*3) #Repetition of string
print(s1+" "+s2) #Concatenation of two strings with space
```

```python
num1=10
print(s1+str(num1)) #Concatenation of string and number
```

```python
name="Nihal Mishra"
print(len(name)) #Length of string
print(name.lower()) #Lower case of string
print(name.upper()) #Upper case of string
print(name.rstrip("ra")) #Removes the trailing characters
print(name.lstrip("Ni")) #Removes the leading characters
print(name.replace("Nihal","Nihal Kumar")) #Replaces the string
print(name.split()) #changing sting to list
print(name.split()) #changing sting to list
```

```python
a="hello how are you,how is your day 1"
print(a.capitalize()) #Capitalize the first letter of string
print(a.center(50)) #Center the string
print(a.endswith("day")) #Check if string ends with the given string
print(a.startswith("hello")) #Check if string starts with the given string
print(a.find("how")) #Find the given string in the string
print(a.islower()) #Check if string is in lower case
print(a.isupper()) #Check if string is in upper case
print(a.istitle()) #Check if string is in title case
print(a.swapcase()) #Swap the case of string
print(a.title()) #Convert the string to title case
```
