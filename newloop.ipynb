loop statement(count,while,for,break,continue,range)

```python
count=1#initialization
while count<=5:#condition
    print("hello Nihal",count)#statement
    count+=1#increment
```

```python
i=1#initialization
while i<=5:#condition
    print(i)#print statement
    i+=1#increment
print("End of the loop")#print statement    
```

```python
i=5#initialization
while i>=1:#condition
    print(i)#print statement
    i-=1#decrement
print("End of the loop")#print statement
```

```python
n=int(input("Enter the number:"))#input statement
i=1#initialization
while i<=10:#condition
    print(n,"*",i,"=",n*i)#print statement
    i+=1#increment
```

```python
nums=[1,4,9,16,25,36,49,64,81,100]#list
idx=0#initialization
while idx<len(nums):#condition
    print(nums[idx])#print statement
    idx+=1#increment
```

```python
i=1#initialization
while i<=5:#condition
    print(i)
    if(i==3):#condition
       break#break statement
    i+=1
```

```python
# Example: Stopping the loop when the number 5 is found
for num in range(1, 10):#range
    if num == 5:#condition
        print("Found 5, breaking the loop!")#print statement
        break
    print(f"Current number: {num}")#print statement
```

```python
name="nihal"#string
for i in name:#loop statement
    print(i)
    if(i=='h'):#condition
        print("Found h, breaking the loop!")
```

```python
colors=["red","green","blue","yellow","black"]#list
for color in colors:#loop statement
    print(color)
    for i in color:#loop statement
        print(i)#print statement
```

```python
for k in range(5):#range
    print(k+1)#print statement
 
```

```python
# Skip even numbers and print only odd numbers from 1 to 10
for num in range(1, 11):
    if num % 2 == 0:  # Check if the number is even
        continue  # Skip the rest of the loop for even numbers
    print(num)
```

```python
# Print numbers from 1 to 10, but skip multiples of 3
for num in range(1, 11):
    if num % 3 == 0:  # Check if the number is a multiple of 3
        continue  # Skip multiples of 3
    print(num)
```
