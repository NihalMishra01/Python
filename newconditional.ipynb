CONDITIOANAL STATEMENT IF-ELSE

```python
i=int(input("enter a number:"))#input function is used to take input from user
if(i%2==0):#if else statement is used to check the condition
    print("even")#print function is used to print the output
else:#else statement is used to check the condition
    print("odd")#print function is used to print the output
```

```python
x = int(input("Enter a number: ")) #input function is used to take input from the user
if x < 5:#if statement is used to check the condition
    print("x is less than 5")#if the condition is true then this statement will be printed
elif x == 10:#elif is used to check another condition if the first condition is false
    print("x is equal to 10")#if the condition is true then this statement will be printed
else:#if the condition is false then this statement will be printed
    print("x is greater than 10")#if the condition is false then this statement will be printed
```

```python
x = int(input("Enter a number: "))#input function is used to take input from the user
if x > 10 and x < 20:#if statement is used to check the condition
    print("x is between 10 and 20")#if the condition is true then this statement will be printed
elif(x>10):
    print("x is smaller than 10")#if the condition is true then this statement will be printed
else:
    print("x is greater than 20") 
       
```

```python
marks = int(input("Enter your marks: "))#input function is used to take input from the user

if marks >= 50:  # Check if the student passed
    print("You passed! 🎉")
    if marks >= 90:#if marks is greater than 90 then this statement will be printed
        print("Grade: A")
    elif marks >= 75:#if marks is greater than 75 then this statement will be printed
        print("Grade: B")
    elif marks >= 60:#if marks is greater than 60 then this statement will be printed
        print("Grade: C")
    else:#if the condition is false then this statement will be printed
        print("Grade: D")
else:#if the condition is false then this statement will be printed
    print("You failed. Better luck next time. 😞")
```
