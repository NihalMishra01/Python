```python
class student:
    def __init__(self, name,hindi_marks,english_marks,science_marks):
        self.name=name
        self.hindi_marks=hindi_marks
        self.english_marks=english_marks
        self.science_marks=science_marks
    @staticmethod
    def welcome():
        print("Hello Nihal")    
        
    def average(self):
        return (self.hindi_marks+self.english_marks+self.science_marks)/3  

s1=student("Nihal",90,80,99)      
s2=student("deepak",90,80,75)
print(s1.average())
print(s2.average())
s1.welcome()
```

```python
class student:
    def __init__(self,name,marks):
        self.name=name
        self.marks=marks
    def average(self):
        sum=0
        for val in self.marks:
            sum+=val
         
        print(f"Hi {self.name} your average is",sum/3)  
s1=student("Nihal",[95,97,99])
s1.average()
s1.name="deepak"
s1.average()
```
